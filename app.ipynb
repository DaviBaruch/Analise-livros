```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('best-selling-books.csv')
```

```python
df.head()
df.info()
```

```python
df['Genre'].isnull().sum()
df['Genre'].fillna('Desconhecido', inplace=True)
```

```python
top10 = df.nlargest(10, 'Approximate sales in millions')
plt.figure(figsize=(12, 6))
sns.barplot(data=top10, y='Book', x='Approximate sales in millions', palette='viridis')
plt.title('Top 10 livros mais vendidos da história')
plt.xlabel('Vendas aproximadas (em milhões)')
plt.ylabel('Livro')
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=df, x='First published', y='Approximate sales in millions', hue='Genre')
plt.title('Vendas x Ano de publicação')
plt.xlabel('Ano de publicação')
plt.ylabel('Vendas (milhões)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
